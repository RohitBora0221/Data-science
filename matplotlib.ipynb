```python
 import matplotlib.pyplot as plt
from matplotlib import colors as mcolors
import numpy as np
import pandas as pd
```

```python
print(mcolors.CSS4_COLORS.values())
```

```python
# #
# #
# Sample car sales company data
data = {
    'Car_Model': ['Sedan', 'SUV', 'Hatchback', 'Coupe', 'Convertible', 'Truck', 'Wagon', 'Van', 'Crossover', 'Hybrid'],
    'Units_Sold': [1500, 2500, 2000, 800, 400, 1200, 900, 700, 1600, 1000],
    'Average_Price': [25000, 35000, 20000, 40000, 45000, 30000, 28000, 27000, 32000, 30000],
    'Customer_Satisfaction': [8.5, 9.2, 8.8, 8.0, 7.8, 8.6, 8.3, 8.0, 8.9, 9.5],
    'Profit_Margin': [15, 20, 18, 12, 10, 14, 16, 13, 19, 25],
    'Yearly_Growth': [5, 7, 6, 4, 3, 5, 4, 3, 6, 8]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Plot 1: Bar chart of units sold per car model
plt.figure(figsize=(10, 6))
bars = plt.bar(df['Car_Model'], df['Units_Sold'], color="green", edgecolor='red', linewidth=1.2, label="Units Sold")
plt.title('Units Sold per Car Model', fontsize=16, fontweight='bold')
plt.gca().set_axisbelow(True)
plt.grid(axis='y', linestyle='--', alpha=1)
plt.xlabel('Car Model', fontsize=14, fontweight='bold')
plt.ylabel('Units Sold', fontsize=14, fontweight='bold')
plt.xticks(rotation=90, fontsize=14)
plt.legend(title='cars')
plt.tight_layout()
plt.show()
```

```python
# Plot 2: line chart of avg price for each  car model
plt.figure(figsize=(10, 6))
plt.plot(df['Car_Model'], df['Average_Price'],marker = '*', color="green", linewidth=1.2, label="Car modal")
plt.title('Average Price per Car Model', fontsize=16, fontweight='bold')
plt.gca().set_axisbelow(True)
plt.grid(axis='y', linestyle='--', alpha=1)
plt.xlabel('Car Model', fontsize=14, fontweight='bold')
plt.ylabel('Average Price', fontsize=14, fontweight='bold')
plt.xticks(rotation=90, fontsize=14)
plt.legend(title='cars')
plt.tight_layout()
plt.show()
```

```python
# Plot 3: scatter chart of customer setisfaction for each  profit margin
plt.figure(figsize=(10, 6))
plt.scatter(df['Customer_Satisfaction'], df['Profit_Margin'], color='skyblue')
plt.title('Customer Satisfaction vs Profit Margin')
plt.xlabel('Customer Satisfaction (Rating)')
plt.ylabel('Profit Margin (%)')
plt.grid(True)
plt.show()
```

```python
# Plot 4: Horizontal bar chart of profit margins
plt.figure(figsize=(10, 6))
plt.barh(df['Car_Model'], df['Profit_Margin'],color='red')
plt.title('Profit Margin per Car Model')
plt.xlabel('Profit Margin (%)')
plt.ylabel('Car Model')
plt.show()
```

```python
# Plot 6: Histogram of yearly growth percentages

plt.figure(figsize=(10, 6))
plt.hist(df['Yearly_Growth'], bins=3, color='green', edgecolor='black')
plt.title('Distribution of Yearly Growth')
plt.xlabel('Yearly Growth (%)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Example list of colors for each slice

custom_colors = ['red', 'blue', 'green']

# Plot the pie chart with explode effect
plt.figure(figsize=(5, 7))
plt.pie(df['Units_Sold'], labels=df['Car_Model'], autopct='%1.1f%%')
plt.title('Car Model Distribution by Units Sold')
plt.show()
#
```
