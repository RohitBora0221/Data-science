```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
sns.set_theme(style="whitegrid")   # Global theme
```

```python
tips = sns.load_dataset("tips")
flights = sns.load_dataset("flights")
titanic = sns.load_dataset("titanic")
iris = sns.load_dataset("iris")
penguins = sns.load_dataset("penguins")

display(tips)
```

# 1. SCATTER PLOT  — Relationship between two continuous variables


```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip",
                hue="sex", size="size", palette="Set1")
plt.title("Scatter Plot — Total Bill vs Tip")
plt.tight_layout()
plt.show()
```

# 2. LINE PLOT  — Trend over time / ordered variable

```python

flights_pivot = flights.pivot(index="month", columns="year", values="passengers")
plt.figure(figsize=(10, 5))
sns.lineplot(data=flights, x="year", y="passengers",
             hue="month", palette="tab20", legend=False)
plt.title("Line Plot — Passengers per Year by Month")
plt.tight_layout()
plt.show()
```

# 3. BAR PLOT  — Compare category means


```python

plt.figure(figsize=(7, 5))
sns.barplot(data=tips, x="day", y="total_bill",
            hue="sex", palette="pastel", ci="sd")
plt.title("Bar Plot — Average Total Bill by Day & Sex")
plt.tight_layout()
plt.show()
```

# 4. COUNT PLOT  — Frequency of categories (like bar chart for counts)



```python
plt.figure(figsize=(7, 5))
sns.countplot(data=titanic, x="class", hue="survived", palette="Set2")
plt.title("Count Plot — Titanic Survival by Class")
plt.tight_layout()
plt.show()
```

# 5. BOX PLOT  — Distribution + outliers per category



```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="day", y="total_bill",
            hue="smoker", palette="coolwarm")
plt.title("Box Plot — Total Bill Distribution by Day & Smoker")
plt.tight_layout()
plt.show()
```

# 6. VIOLIN PLOT  — Box plot + KDE (smooth distribution shape)



```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=tips, x="day", y="total_bill",
               hue="sex", split=True, palette="muted")
plt.title("Violin Plot — Total Bill by Day & Sex")
plt.tight_layout()
plt.show()
```

# 7. HISTOGRAM  — Distribution of a single numeric variable


```python

plt.figure(figsize=(7, 5))
sns.histplot(data=tips, x="total_bill", bins=20,
             kde=True, color="steelblue")
plt.title("Histogram — Total Bill Distribution")
plt.tight_layout()
plt.show()
```

# 8. KDE PLOT  — Smooth probability density estimate


```python
plt.figure(figsize=(7, 5))
sns.kdeplot(data=tips, x="total_bill", hue="sex",
            fill=True, alpha=0.4, palette="Set1")
plt.title("KDE Plot — Total Bill Density by Sex")
plt.tight_layout()
plt.show()
```

# 9. ECDF PLOT  — Empirical Cumulative Distribution Function


```python
plt.figure(figsize=(7, 5))
sns.ecdfplot(data=tips, x="total_bill", hue="day")
plt.title("ECDF Plot — Cumulative Distribution of Total Bill")
plt.tight_layout()
plt.show()
```

# 10. HEATMAP  — Correlation matrix / pivot table visualization


```python
plt.figure(figsize=(8, 6))
corr = iris.drop("species", axis=1).corr()
sns.heatmap(corr, annot=True, fmt=".2f",
            cmap="coolwarm", linewidths=0.5)
plt.title("Heatmap — Iris Feature Correlations")
plt.tight_layout()
plt.show()
```

# 11. PAIR PLOT  — All pairwise scatter plots + distributions



```python
pair_grid = sns.pairplot(iris, hue="species",
                         palette="husl", diag_kind="kde")
pair_grid.figure.suptitle("Pair Plot — Iris Dataset", y=1.02)
plt.show()
```

# 12. REG PLOT  — Scatter + linear regression line


```python
plt.figure(figsize=(7, 5))
sns.regplot(data=tips, x="total_bill", y="tip",
            scatter_kws={"alpha": 0.5}, line_kws={"color": "red"})
plt.title("Reg Plot — Total Bill vs Tip with Regression Line")
plt.tight_layout()
plt.show()
```

# 13. LM PLOT  — regplot with FacetGrid (multiple subplots)



```python
lm = sns.lmplot(data=tips, x="total_bill", y="tip",
                col="sex", hue="smoker", palette="Set1",
                height=5, aspect=0.8)
lm.figure.suptitle("LM Plot — Regression by Sex & Smoker", y=1.02)
plt.show()
```

# 14. RESIDUAL PLOT  — Check regression residuals


```python
plt.figure(figsize=(7, 5))
sns.residplot(data=tips, x="total_bill", y="tip",
              lowess=True, color="purple")
plt.title("Residual Plot — Regression Residuals")
plt.tight_layout()
plt.show()
```

# 15. STRIP PLOT  — Scatter of categorical vs numeric (show all points)


```python
plt.figure(figsize=(7, 5))
sns.stripplot(data=tips, x="day", y="total_bill",
              hue="sex", jitter=True, dodge=True, palette="Set2")
plt.title("Strip Plot — Total Bill by Day")
plt.tight_layout()
plt.show()
```

# 16. SWARM PLOT  — Strip plot without overlapping points

```python
plt.figure(figsize=(7, 5))
sns.swarmplot(data=tips, x="day", y="total_bill",
              hue="sex", dodge=True, palette="Set1")
plt.title("Swarm Plot — Total Bill by Day")
plt.tight_layout()
plt.show()
```

# 17. POINT PLOT  — Category means with confidence intervals


```python
plt.figure(figsize=(7, 5))
sns.pointplot(data=tips, x="day", y="total_bill",
              hue="sex", palette="deep", dodge=True,
              capsize=0.1, markers=["o", "s"])
plt.title("Point Plot — Mean Total Bill by Day & Sex")
plt.tight_layout()
plt.show()
```

# 18. CLUSTER MAP  — Hierarchical clustering heatmap


```python
flights_matrix = flights.pivot(index="month", columns="year",
                               values="passengers")
sns.clustermap(flights_matrix, cmap="YlOrRd",
               standard_scale=1, figsize=(10, 8))
plt.suptitle("Cluster Map — Flights Passengers (Clustered)", y=1.02)
plt.show()
```

# 19. JOINT PLOT  — Bivariate + marginal distributions


```python
jp = sns.jointplot(data=tips, x="total_bill", y="tip",
                   kind="reg", color="teal",
                   marginal_kws={"bins": 20})
jp.figure.suptitle("Joint Plot — Total Bill vs Tip", y=1.02)
plt.show()

# Joint plot with KDE
jp2 = sns.jointplot(data=tips, x="total_bill", y="tip",
                    kind="kde", fill=True, cmap="Blues")
jp2.figure.suptitle("Joint Plot (KDE) — Total Bill vs Tip", y=1.02)
plt.show()
```

# 20. FACET GRID  — Multiple subplots by categories


```python
g = sns.FacetGrid(tips, col="time", row="sex", height=4)
g.map_dataframe(sns.histplot, x="total_bill", bins=15, kde=True)
g.add_legend()
g.figure.suptitle("Facet Grid — Total Bill by Time & Sex", y=1.02)
plt.show()
```

# 21. CAT PLOT  — General categorical plot factory



```python
g = sns.catplot(data=tips, x="day", y="total_bill",
                col="time", kind="box", palette="Set3",
                height=5, aspect=0.8)
g.figure.suptitle("Cat Plot — Box by Day & Time", y=1.02)
plt.show()
```

# 22. REL PLOT  — Relational plot factory (scatter / line)



```python
g = sns.relplot(data=tips, x="total_bill", y="tip",
                col="time", hue="sex", style="smoker",
                kind="scatter", palette="Set1",
                height=5, aspect=0.8)
g.figure.suptitle("Rel Plot — Scatter by Time", y=1.02)
plt.show()
```

# 23. DIST PLOT (displot)  — Unified distribution plot



```python
g = sns.displot(data=tips, x="total_bill", col="sex",
                hue="smoker", kind="hist",
                kde=True, palette="dark", height=5)
g.figure.suptitle("Dist Plot — Bill Distribution by Sex & Smoker", y=1.02)
plt.show()
```

# 24. 2D KDE PLOT  — Density of two continuous variables

```python
plt.figure(figsize=(7, 5))
sns.kdeplot(data=tips, x="total_bill", y="tip",
            cmap="Blues", fill=True, thresh=0.05)
plt.title("2D KDE Plot — Total Bill vs Tip Density")
plt.tight_layout()
plt.show()
```

# 25. RUGPLOT  — Marginal ticks along axis

```python
plt.figure(figsize=(7, 5))
sns.kdeplot(data=tips, x="total_bill", fill=True, color="steelblue")
sns.rugplot(data=tips, x="total_bill", color="red", height=0.05)
plt.title("Rug Plot — Total Bill KDE + Rug Ticks")
plt.tight_layout()
plt.show()
```

# 26. PAIR GRID (custom)  — Fine-grained control over pair plots



```python
g = sns.PairGrid(penguins.dropna(), hue="species")
g.map_upper(sns.scatterplot, alpha=0.5)
g.map_lower(sns.kdeplot, fill=True)
g.map_diag(sns.histplot, kde=True)
g.add_legend()
g.figure.suptitle("Pair Grid — Penguins Dataset", y=1.02)
plt.show()
```

# 27. BOXEN PLOT (Letter-Value Plot)  — Better box plot for big data


```python
plt.figure(figsize=(8, 5))
sns.boxenplot(data=tips, x="day", y="total_bill",
              hue="sex", palette="pastel")
plt.title("Boxen Plot — Total Bill by Day & Sex")
plt.tight_layout()
plt.show()
```
